# 🎥 eidcloud-video-pipeline Quickstart

Welcome to **eidcloud-video-pipeline**, a modular video translation, dubbing, and subtitle orchestration pipeline with audio ducking and FFmpeg in pure PHP 8.2+.

This 3-cell notebook demonstrates:
1. Environment Setup & Generating a Synthetic Audio/Video Clip
2. Running the Audio Extraction & Ducking Pipeline via CLI
3. Multiplexing Synchronized Subtitles into Video

```python
# Cell 1: Environment Setup & Test Media Generation
!apt-get update -qq && apt-get install -y -qq php-cli php-curl ffmpeg
!php -v && ffmpeg -version | head -n 1

# Generate 5-second synthetic test video with audio tone
!ffmpeg -y -f lavfi -i testsrc=duration=5:size=640x360:rate=30 -f lavfi -i sine=frequency=440:duration=5 -pix_fmt yuv420p test_input.mp4 -loglevel quiet
print("✅ Synthetic test video created: test_input.mp4")
```

```python
# Cell 2: Extract Audio & Run Modular Translation Stages
!php bin/eidcloud-video extract-audio test_input.mp4 --out=extracted_audio.wav
!ls -lh extracted_audio.wav

# Run video translation and dubbing
!php bin/eidcloud-video translate test_input.mp4 --from=en --to=ar --voice=male
```

```python
# Cell 3: Audio Ducking & Subtitle Multiplexing
!php bin/eidcloud-video mix test_input.mp4 extracted_audio.wav --ducking=-12dB
print("\n🎬 Pipeline execution completed successfully!")
```
